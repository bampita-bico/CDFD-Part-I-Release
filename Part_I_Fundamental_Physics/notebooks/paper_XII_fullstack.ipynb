# Paper XII - Full-stack supplementary reproduction

Calls `run_paper_xii()` from `Part_I_Fundamental_Physics/notebooks/supplementary_XII.py` and regenerates the Paper XII figures, tables, and checks.

```python
import os
import sys
from pathlib import Path

HERE = Path.cwd().resolve()
search_roots = [HERE, *HERE.parents]
candidates = []
for base in search_roots:
    candidates.append(base)
    candidates.append(base / "Part_I_Fundamental_Physics")
    candidates.append(base / "Part_I_Fundamental_Physics" / "notebooks")

PAPERS = None
for candidate in candidates:
    if (candidate / "supplementary_XII.py").exists():
        PAPERS = candidate
        break

if PAPERS is None:
    raise RuntimeError("Could not find Part_I_Fundamental_Physics/notebooks/supplementary_XII.py")

ROOT = PAPERS.parent
os.chdir(PAPERS)
sys.path.insert(0, str(PAPERS))
sys.path.insert(0, str(ROOT))

from supplementary_XII import run_paper_xii

results = run_paper_xii(write_outputs=True, verbose=True)
results["output_dir"]
```
