# Paper VI - Full-stack supplementary reproduction

Calls `run_paper_vi()` from `Part_II_Fundamental_Physics/supplementary_VI.py` and regenerates the Paper VI figures, tables, and checks.

```python
import os
import sys
from pathlib import Path

HERE = Path.cwd().resolve()
search_roots = [HERE, *HERE.parents]
candidates = []
for base in search_roots:
    candidates.append(base)
    candidates.append(base / "Part_II_Fundamental_Physics")

PAPERS = None
for candidate in candidates:
    if (candidate / "supplementary_VI.py").exists():
        PAPERS = candidate
        break

if PAPERS is None:
    raise RuntimeError("Could not find Part_II_Fundamental_Physics/supplementary_VI.py")

ROOT = PAPERS.parent
os.chdir(PAPERS)
sys.path.insert(0, str(PAPERS))
sys.path.insert(0, str(ROOT))

from supplementary_VI import run_paper_vi

results = run_paper_vi(write_outputs=True, verbose=True)
results["output_dir"]
```
